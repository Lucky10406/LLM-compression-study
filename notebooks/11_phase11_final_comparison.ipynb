```python
summary_cols = [
    "model",
    "accuracy",
    "macro_f1",
    "weighted_f1",
    "median_latency_s",
    "peak_gpu_memory_gb",
    "energy_100_kwh",
    "checkpoint_size_gb",
    "usable"
]

print(
    results_df[summary_cols]
    .to_string(index=False)
)
```

```python
baseline = results_df[
    results_df["model"] == "FP16_Qwen"
].iloc[0]

usable_df = results_df[
    results_df["usable"] == True
].copy()

usable_df["accuracy_delta_pp"] = (
    usable_df["accuracy"]
    - baseline["accuracy"]
) * 100

usable_df["macro_f1_delta"] = (
    usable_df["macro_f1"]
    - baseline["macro_f1"]
)

usable_df["memory_reduction_pct"] = (
    1
    -
    usable_df["peak_gpu_memory_gb"]
    / baseline["peak_gpu_memory_gb"]
) * 100

usable_df["energy_reduction_pct"] = (
    1
    -
    usable_df["energy_100_kwh"]
    / baseline["energy_100_kwh"]
) * 100

print(
    usable_df[
        [
            "model",
            "accuracy_delta_pp",
            "macro_f1_delta",
            "memory_reduction_pct",
            "energy_reduction_pct"
        ]
    ].to_string(index=False)
)
```

```python
pareto_df = results_df[
    (results_df["usable"] == True)
    &
    (results_df["family"] != "ablation")
].dropna(
    subset=[
        "macro_f1",
        "median_latency_s",
        "peak_gpu_memory_gb",
        "energy_100_kwh"
    ]
).copy()

print(pareto_df["model"].tolist())
```

```python
def is_dominated(
    row,
    df,
    metrics_to_maximize,
    metrics_to_minimize
):
    for _, other in df.iterrows():

        if other["model"] == row["model"]:
            continue

        better_or_equal = True
        strictly_better = False

        for metric in metrics_to_maximize:

            if other[metric] < row[metric]:
                better_or_equal = False
                break

            if other[metric] > row[metric]:
                strictly_better = True

        if not better_or_equal:
            continue

        for metric in metrics_to_minimize:

            if other[metric] > row[metric]:
                better_or_equal = False
                break

            if other[metric] < row[metric]:
                strictly_better = True

        if better_or_equal and strictly_better:
            return True

    return False
```

```python
metrics_to_maximize = [
    "macro_f1"
]

metrics_to_minimize = [
    "median_latency_s",
    "peak_gpu_memory_gb",
    "energy_100_kwh"
]

pareto_df["dominated"] = pareto_df.apply(
    lambda row: is_dominated(
        row,
        pareto_df,
        metrics_to_maximize,
        metrics_to_minimize
    ),
    axis=1
)

print(
    pareto_df[
        [
            "model",
            "macro_f1",
            "median_latency_s",
            "peak_gpu_memory_gb",
            "energy_100_kwh",
            "dominated"
        ]
    ].to_string(index=False)
)
```

```python
collapsed = results_df[
    results_df["usable"] == False
]

print(
    collapsed[
        [
            "model",
            "accuracy",
            "macro_f1",
            "unmatched_rate",
            "median_latency_s",
            "energy_100_kwh"
        ]
    ].to_string(index=False)
)
```

```python
distilled = results_df[
    results_df["model"]
    == "Distilled_DistilBERT"
].iloc[0]

hard = results_df[
    results_df["model"]
    == "HardLabel_DistilBERT"
].iloc[0]

print(
    "Accuracy gain:",
    round(
        (distilled["accuracy"]
         - hard["accuracy"]) * 100,
        3
    ),
    "percentage points"
)

print(
    "Macro F1 gain:",
    round(
        distilled["macro_f1"]
        - hard["macro_f1"],
        6
    )
)

print(
    "Weighted F1 gain:",
    round(
        distilled["weighted_f1"]
        - hard["weighted_f1"],
        6
    )
)
```
